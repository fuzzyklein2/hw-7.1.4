## Make a docstring

```python
def doxify(text):
    # Just add comment delimiters and asterisks.
    lines = text.split('\n')[1:]
    result = ['/**']
    result.extend([' * ' + s for s in lines])
    result.append(' */')
    return '\n'.join(result)
```

```python
s = (doxify("""
Process paths.



@param target The substring to find and replace.
@param replacement String to replace the target with.
@return Copy of `this` with all occurences of `target` replaced by `replacement`."""))
print(s)
pyperclip.copy(s)
print("Docstring copied to clipboard")
```
