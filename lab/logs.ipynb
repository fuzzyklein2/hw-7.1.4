## View the latest log file

```python
DATA_FILE = "data/data.json"
with Path(DATA_FILE).open() as f:
    DATA = json.load(f)
PROGRAM = DATA["program"]
logs = Path.home() / f".log/{PROGRAM}"
files = sorted(os.listdir(logs))
f = Path(logs) / files[-1]
print(f"File: {f}")
print()
print(f.read_text())
```
