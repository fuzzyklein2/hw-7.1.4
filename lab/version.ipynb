## Update Repository

````python
MESSAGE = "Test git from Jupyter."
GIT = 'git'
OPTIONS = {
    "text" : True,
    "capture_output" : True,
    "check" : False,
    "shell" : False
}
process = subprocess.run([
        GIT, "add", "."
    ],
        **OPTIONS
    )
print(dir(process))
print("Added all files to commit.")
process = subprocess.run([
        GIT, "commit", "-m", '"'+MESSAGE+'"'
    ], **OPTIONS
    )
print("Changes committed.")
process = subprocess.run([
        GIT, "push"
    ], **OPTIONS
    )
print("Repository pushed to GitHub.")
````
